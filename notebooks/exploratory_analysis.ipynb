# Exploratory Spatial Justice Analysis: Chicago Library Accessibility

This supplementary notebook provides interactive visualizations and exploratory data inspection for the Chicago spatial access research project.

**Core Research Question:** Are higher-poverty census tracts in Chicago associated with longer walking distances to public libraries?

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Load tract-level metrics dataset
df = pd.read_csv('../results/tables/chicago_tract_level_metrics.csv')
print(f"Loaded {len(df)} populated Chicago census tracts.")
df[['poverty_rate', 'network_distance_m', 'access_score_1000', 'pop_density_km2']].describe()
```

```python
# Compute Spearman rank correlation
rho, p_val = stats.spearmanr(df['poverty_rate'], df['network_distance_m'])
print(f"Spearman rank correlation: rho = {rho:.4f}, p-value = {p_val:.4e}")
```

```python
# Poverty Rate Distribution vs Walking Distance
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df['poverty_rate'], df['network_distance_m'] / 1000.0, alpha=0.5, c='#2b83ba', edgecolors='none')
ax.set_xlabel('Tract Poverty Rate (q_i)')
ax.set_ylabel('Walking Distance to Nearest Library (km)')
ax.set_title('Tract Poverty Rate vs. Nearest Library Distance')
ax.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
